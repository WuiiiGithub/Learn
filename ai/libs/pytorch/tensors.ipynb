```python
import torch
import numpy as np
```

```python
data = [
    [
        [2, 3],
        [1, 4]
    ],
    [
        [4, 3],
        [3, 2]
    ],
    [
        [4, 6],
        [1, 2]
    ]
]
tensorX = torch.tensor(data)
tensorX
```

```python
tensorX_np = torch.from_numpy(np.array(data))
tensorX_np == tensorX
```

```python
print("torch.ones:\n---------")
print(torch.ones(size=(2, 3)))

print("\n=========\n")

print("torch.ones_like:\n---------")
print(torch.ones_like(input=tensorX))
```

```python
print("torch.zeros:\n---------")
print(torch.zeros(size=(2, 3)))

print("\n=========\n")

print("torch.zeros_like:\n---------")
print(torch.zeros_like(input=tensorX))
```

```python
print("torch.rand:\n---------")
print(torch.rand(size=(2, 3)))

print("\n=========\n")

print("torch.rand_like:\n---------")
print(torch.rand_like(input=tensorX, dtype=torch.float))
```

```python
print("Fetching 2nd element of each array")
tensorX[:, :, 1]
```

```python
print("Other way to fetch 2nd element of each array")
tensorX[..., 1]
```

```python
print("Actual Array:\n------")
print(tensorX)

print("\n==========\n")

print("Concatinated (Horizontally):\n-------")
print(tconc:=torch.concat([
    tensorX,
    torch.zeros_like(tensorX),
    torch.ones_like(tensorX)
]))
print("Shape is", tconc.shape)

print("\n==========\n")

print("Stacked (Vertically):\n-------")
print(tstack:=torch.stack([
    tensorX,
    torch.zeros_like(tensorX),
    torch.ones_like(tensorX)
]))
print("Shape is", tstack.shape)
```
