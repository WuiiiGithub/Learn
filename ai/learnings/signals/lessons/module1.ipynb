## Visualising Basic Waveforms

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
time = np.linspace(-3, 3, 10000)
sine_wave = np.sin(2 * np.pi * 1 * time)

plt.scatter(time, sine_wave, marker='.')
plt.yticks([-1, -0.75, -0.5, -0.25, 0, 0.25, 0.5, 0.75, 1])
plt.xlabel("time")
plt.ylabel("sin(2$\\pi$ft)")
plt.title("A sine wave")
plt.show()
```

```python
time = np.linspace(-3, 3, 10000)
cos_wave = np.cos(2 * np.pi * 1 * time)

plt.scatter(time, cos_wave, marker='.')
plt.yticks([-1, -0.75, -0.5, -0.25, 0, 0.25, 0.5, 0.75, 1])
plt.xlabel("time")
plt.ylabel("cos(2$\\pi$ft)")
plt.title("A cos wave")
plt.show()
```

```python
time = np.linspace(-3, 3, 10000)
sine_wave = 0.25*np.sin(2 * np.pi * 1 * time)


plt.scatter(time, sine_wave, marker='.')
plt.yticks([-1, -0.75, -0.5, -0.25, 0, 0.25, 0.5, 0.75, 1])
plt.xlabel("time")
plt.ylabel("sin(2$\\pi$ft)")
plt.title("A sine wave")
plt.show()
```

```python
time = np.linspace(-3, 3, 10000)
sine_wave = 0.25*np.sin(2 * np.pi * 1 * time) + 0.5


plt.scatter(time, sine_wave, marker='.')
plt.yticks([-1, -0.75, -0.5, -0.25, 0, 0.25, 0.5, 0.75, 1])
plt.xlabel("time")
plt.ylabel("sin(2$\\pi$ft)")
plt.title("A sine wave")
plt.show()
```

```python
time = np.linspace(-3, 3, 10000)
sine_wave = np.sin(2 * np.pi * 1 * time)
square_wave = np.sign(sine_wave)

plt.scatter(time, square_wave, marker='.')
plt.yticks([-1, -0.75, -0.5, -0.25, 0, 0.25, 0.5, 0.75, 1])
plt.xlabel("time")
plt.ylabel("sign(sin(2$\\pi$ft))")
plt.title("A square wave")
plt.show()
```

```python
time = np.linspace(-3, 3, 10000)
sine_wave = np.sin(2 * np.pi * 1 * time)
square_wave = np.sign(sine_wave)

plt.plot(time, square_wave)
plt.yticks([-1, -0.75, -0.5, -0.25, 0, 0.25, 0.5, 0.75, 1])
plt.xlabel("time")
plt.ylabel("sign(sin(2$\\pi$ft))")
plt.title("A square wave")
plt.show()
```

```python
shifted = np.roll(sine_wave, 10000//12)
scaled = 2 * sine_wave

plt.figure(figsize=(12, 4))
plt.plot(time, sine_wave, label='Original')
plt.plot(time, shifted, label='Time-Shifted')
plt.plot(time, scaled, label='Scaled (2x)')
plt.legend()
plt.title('Time & Amplitude Transformations')
plt.grid(True)
plt.show()
```

```python
transformed = np.flip(sine_wave)

plt.figure(figsize=(12, 4))
plt.plot(time, sine_wave, label='Original')
plt.plot(time, transformed, label='Time-Reversed')
plt.legend()
plt.title('Time Reversal Transformations')
plt.grid(True)
plt.show()
```

```python
transformed = np.clip(sine_wave, -0.5, 0.5)

plt.figure(figsize=(12, 4))
plt.plot(time, sine_wave, label='Original')
plt.plot(time, transformed, label='Clipped Transform')
plt.legend()
plt.title('Clipped Transformations')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(12, 10))

plt.subplot(4, 1, 1)
plt.plot(time, sine_wave, label='Original', color='lightgreen')
plt.plot(time[::50], sine_wave[::50], '--', color='orange', linewidth=2, label='Signal with 50')
plt.plot(time[::50], sine_wave[::50], 'o', color='red', markersize=5)
plt.title('Downsample x50')
plt.grid(True)
plt.legend()

plt.subplot(4, 1, 2)
plt.plot(time, sine_wave, label='Original', color='lightgreen')
plt.plot(time[::100], sine_wave[::100], '--', color='orange', linewidth=2, label='Signal with 100')
plt.plot(time[::100], sine_wave[::100], 'o', color='red', markersize=5)
plt.title('Downsample x100')
plt.grid(True)
plt.legend()

plt.subplot(4, 1, 3)
plt.plot(time, sine_wave, label='Original', color='lightgreen')
plt.plot(time[::500], sine_wave[::500], '--', color='orange', linewidth=2, label='Signal with 500')
plt.plot(time[::500], sine_wave[::500], 'o', color='red', markersize=5)
plt.title('Downsample x500')
plt.grid(True)
plt.legend()

plt.subplot(4, 1, 4)
plt.plot(time, sine_wave, label='Original', color='lightgreen')
plt.plot(time[::1000], sine_wave[::1000], '--', color='orange', linewidth=2, label='Signal with 1000')
plt.plot(time[::1000], sine_wave[::1000], 'o', color='red', markersize=5)
plt.title('Downsample x1000')
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```

# Attempted Frequency Transform

```python
linear_time_func = lambda m,c=0: m*time + c

plt.subplot(2, 1, 1)
plt.plot(
    time,
    linear_time_func(1)
)
plt.title("Frequency in Linear Time")
plt.ylabel("Frequency with time")

plt.subplot(2, 1, 2)
plt.plot(
    time,
    np.sin(2 * np.pi * linear_time_func(1))
)
plt.xlabel("Linear Time")
plt.ylabel("Signal")

plt.tight_layout()
plt.show()
```

```python
quadratic_time_func = lambda a,b,c=0: a*time**2 + b*time + c

plt.subplot(2, 1, 1)
plt.plot(
    time,
    quadratic_time_func(1, 0)
)
plt.title("Frequency in Quadratic Time")
plt.ylabel("Frequency with time")

plt.subplot(2, 1, 2)
plt.plot(
    time,
    np.sin(2 * np.pi * quadratic_time_func(1, 0))
)
plt.xlabel("Quadratic Time")
plt.ylabel("Signal")

plt.tight_layout()
plt.show()
```

```python
cubic_time_func = lambda a,b,c,d=0: a*time**3 + b*time**2 + c*time + d

plt.subplot(2, 1, 1)
plt.plot(
    time,
    cubic_time_func(1, 0, 0)
)
plt.title("Frequency in Cubic Time")
plt.ylabel("Frequency with time")

plt.subplot(2, 1, 2)
plt.plot(
    time,
    np.sin(2 * np.pi * cubic_time_func(1, 0, 0))
)
plt.xlabel("Cubic Time")
plt.ylabel("Signal")

plt.tight_layout()
plt.show()
```

```python
time2 = np.linspace(-5, 5, 5000)

gaussian_func = lambda mu=0, sigma=1: (
    np.exp(-((time2 - mu)**2) / (2 * sigma**2))
)

freq = gaussian_func(mu=0, sigma=1)

plt.subplot(2, 1, 1)
plt.plot(time2, freq)
plt.title("Gaussian Frequency vs Time")
plt.ylabel("Frequency")

plt.subplot(2, 1, 2)
plt.plot(
    time2,
    np.sin(2 * np.pi * freq)
)
plt.xlabel("Time")
plt.ylabel("Signal")

plt.tight_layout()
plt.show()
```

```python
time = np.linspace(-5, 5, 2000)

bell_curve = lambda mu=0, sigma=1: (
    np.exp(-((time - mu)**2)/(2*sigma**2))
)

plt.subplot(2,1,1)
plt.plot(time, bell_curve())
plt.title("Bell Curve vs Time")
plt.ylabel("Value")

plt.subplot(2,1,2)
plt.plot(
    time,
    np.sin(2*np.pi*bell_curve())
)
plt.xlabel("Time")
plt.ylabel("Signal")

plt.tight_layout()
plt.show()
```

# Amplitude Trnasform

```python
time = np.linspace(-3, 3, 10000)
sine_wave = np.sin(time)
```

```python
linear_time_func = lambda m, c=0: m * time + c

plt.subplot(2, 1, 1)
plt.plot(time, linear_time_func(1))
plt.title("Amplitude Envelope in Linear Time")
plt.ylabel("Amplitude with time")

plt.subplot(2, 1, 2)
envelope = linear_time_func(1)
signal = np.multiply(envelope, np.sin(2 * np.pi * 1 * time))

plt.plot(time, signal, label="Signal")
plt.plot(time, envelope, 'r--', label="Upper Bound (+)")
plt.plot(time, -envelope, 'b--', label="Lower Bound (-)")

plt.xlabel("Linear Time")
plt.ylabel("Signal")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
quadratic_time_func = lambda a,b,c=0: a*time**2 + b*time + c

plt.subplot(2, 1, 1)
plt.plot(time, quadratic_time_func(1, -2))
plt.title("Amplitude Envelope in Linear Time")
plt.ylabel("Amplitude with time")

plt.subplot(2, 1, 2)
envelope = quadratic_time_func(1, -2)
signal = np.multiply(envelope, np.sin(2 * np.pi * 1 * time))

plt.plot(time, signal, label="Signal")
plt.plot(time, envelope, 'r--', label="Upper Bound (+)")
plt.plot(time, -envelope, 'b--', label="Lower Bound (-)")

plt.xlabel("Quadratic Time")
plt.ylabel("Signal")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
cubic_time_func = lambda a,b,c,d=0: a*time**3 + b*time**2 + c*time + d

plt.subplot(2, 1, 1)
plt.plot(
    time,
    cubic_time_func(1, 0, 0)
)
plt.title("Frequency in Cubic Time")
plt.ylabel("Frequency with time")

plt.subplot(2, 1, 2)
plt.plot(
    time,
    np.sin(2 * np.pi * cubic_time_func(1, 0, 0))
)
plt.xlabel("Cubic Time")
plt.ylabel("Signal")

plt.tight_layout()
plt.show()
```
